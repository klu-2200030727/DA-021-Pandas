# Decorators

```python
# Def: A Decorator is a function that makes another function as input and 
#      add the EXTRA functionality without changing the orignal function code.
```

```python
# Example - 1
def decorator_func(cake):
    def wrapper_fun():
        print("Buy ingredient for cake")      # Extra functionality - 1
        cake()
        print("Put the cherry on the top")    # Extra functionality - 2
    return wrapper_fun    
        
def cake():    # Actual function which you want to decorate  
    print("Take wheat Flour")
    print("Mix some sugar and milk")
    print("Add Egg and baking soda")
    print("Put in the oven")
updated_func = decorator_func(cake)    #passing the actual function to decorator function
updated_func()  # decorated function
```

# The @ symbol for defining decorators

```python
def decorator_func1(dish):        #decorator function which will recieve the actual function
    def wrapper_fun():           # wrapper function will add extra functionality to the actual function 
        
        dish()
        print("put the cherry on the top")# Extra functionality -2
        print("add coco")
        
    return wrapper_fun 
@decorator_func1
def cake():    # Actual function which you want to decorate
    print("Take wheat Flour")
    print("mix some sugar and milk")
    print("add egg and baking soda")
    print("put in the oven")
cake()
```

# Python Encapsulation

```python
# Def : Encapsulation refers to the binding of data and functions into a single unit is called Encapsulation
# Due to te encapsulation, only the same class/subclass functions can access the variables of that class.
# Type of attributes/variables in encapsulation 
# 1. Public
# 2. Protected
# 3. Private
```

```python
# Example
class marvels:
    def __init__(self,name,power,weapon):
        self.name = name
        self.power = power            # instance variable : this is public variable
        self.weapon = weapon
    def display(self):
        print(self.name,self.power,self.weapon)
obj1 = marvels("Thor",10000, "strome")
obj2 = marvels("Hulk", 20000, "fist")
print(obj1.name)
print(obj2.name)
```

```python
# Example   : public
class marvels:
    def __init__(self,name,power,weapon):
        self.Name = name   #instance variable   : this is public variable
        self._power = power   #instance variable   : this is protected variable
        self.__weapon = weapon  #instance variable  : this is private variable
    def display(self):
        print(self.Name, self._power,self.__weapon)
obj1 = marvels("thor","10000","strom breaker")      
obj2 = marvels("Hulk","1000","Fist")   

print(obj1.Name)


obj1.display()
```

```python
# Example - 2 : revise
class Nanna:
    def __init__(self,vehicle,Money, property):
        self.vehicle = vehicle   # Public variable : can be access outside
        self._Money = Money       #  Protected  : can access outside but not suppose to
        self.__property = property
obj = Nanna("Range Rover", 5000000, 1000000)
print(obj.vehicle)
print(obj._Money)
print(obj._Nanna__property)
```

```python
# Example -2 : add new member fynction in base class
# Parent class/Base class 
class Hyd_food:
    
    taste = "very very very spicy"

    def biryani(self):
        print("hyd biryani is very very very tasty")
        
    def shawarma(self):
        print("veg shawarma")
# child class / derived class
class Mehfil(Hyd_food):
    def display(self):
        print("i am in child class")
        
# With parent class object we cannot access the child class function and variables
# But with child class object we can access the parent class functions and variables 
obj_parent = Hyd_food()
print(obj_parent.taste)
obj_parent.biryani()
obj_parent.shawarma()

obj = Mehfil()
print(obj.taste)
obj.biryani()
obj.shawarma()
obj.display()
```

```python
# Example - 3 super()
# super () is used to access parent or grandparent functions and attributes
class Hyd_food:
    
    taste = "very very very spicy"

    def Biryani(self):
        print("Hyd biryani is very very very tasty")
        
    def shawarma(self):
        print("veg shawarma")
# child class / derived class
class Mehfil(Hyd_food):
    def display(self):
        print("i am in child class")
        super().Biryani()
        super().shawarma()
        print(super().taste)
obj_mehfil = Mehfil()
obj_mehfil.display()
```

# 1. Function overloading

```python
# How function overloading works  in C++/Java

def fun(var1):
    print(var1)
def fun(var1, var2):
    print(var1, var2)
fun(10)
fun(100,200)
```

```python
# using default arguments

def fun(var1, var2 = None):
    if var2 is not None:
        print(var1, var2)
    else:
        print(var1)
fun(10)
fun(100,200)

# Here fun() function is overloaded as it can process single value also
# and multiple values also
```

# Arithmetic operator overloading

```python
class Dosa:
    def __init__(self,variety,price):
        self.variety  = variety
        self.price = price
    def __add__(self,second_obj):
        return self.price + second_obj.price
    def __sub__(self,second_obj):
        return self.price - second_obj.price
    def __mul__(self,second_obj):
        return self.price * second_obj.price
    def __truediv__(self,second_obj):
        return self.price / second_obj.price 
    def __floordiv__(self,second_obj):
        return self.price / second_obj.price    
d1 = Dosa("Masala Dosa", 50)
d2 = Dosa("Egg Dosa", 75)
print(d1 + d2)
print(d1 - d2)
print(d1 * d2)
print(d2 / d1)
print(d1 // d2)     
```

# Data Abstraction

```python
# Def : Abstraction hides internal implementation details and shows only essential features.
# Eg: ATM hides the unnecessary details from the user and allow to enter only pincode, amount.
# Benfits of Abstraction:
# 1. Improve the code readability, by hiding complex details
# 2. Privacy & Security
# Library used for data abstraction 
# From abc import ABC, abstract method


# Rules of Abstraction:
  # 1. Abstract class can have both abstract and concrete method.
  # 2. You Cannot instantiate (create object) an Abstract class.
  # 3. All subclasses/child class must implement all abstract methods.

# Abstract Method: A method which is declared but not need to  defined / implemented.
# Concrete method : This is fully implemented method defined in abstract class.
```

```python
# Example - 1 : Structure of abstract class, abstract method

from abc import ABC, abstractmethod   # ABS is a class, abstractmethod is a decorator

# Absract class: This will store hiding information.

class TATA_main_branch(ABC):
    @abstractmethod
    def Air_bag(self):
        pass

class Hyd_TATA(TATA_main_branch):
    def Air_bag(self):
        print("Designing Air_bag with cotton")

obj_hyd = Hyd_TATA()
obj_hyd.Air_bag()
        
```

```python
# re.findall(): returns all non-overlapping matches as a list.

txt = '''After Exam result out 
Mom : Koduku , see the Son of sharma Ji
Son : sorry Mom, I only focused on his Daughtor
'''

findall_obj = re.findall("Son",txt)
print(findall_obj)
```

```python
# re.split() : split the text based upon a pattern(\n, \t, " ", ",") and return a list

split_obj = re.split("\n",txt)
print(split_obj)
```

```python
# re.split() : split the text based upon a pattern(\n, \t, " ", ",") and return a list

split_obj = re.split(" ",txt)
print(split_obj)
```

```python
# re.split() : split the text based upon a pattern(\n, \t, " ", ",") and return a list

split_obj = re.split(",",txt)
print(split_obj)
```

```python
# re.compile() : precomplie a pattern

# [A-Z][a-z] + : this pattern is used to search strings with capital case letter
pattern = re.compile("[A-Z][a-z]+")
findall_obj = pattern.findall(txt)
print(findall_obj)
```

# Generators

```python
# In Generators, yield keyword is used.
# Generators mainly used with the functions in python 
```

```python
def fun():
    i = 10
    j = 20
    return i+j
    print("Nikhil")
print(fun())    
```

```python
def fun():
    i = 10
    j = 20
    yield i+j
    print("Hello")
    yield i*j
    print("World")
    yield i/j

gen_obj = iter(fun())

print(next(gen_obj))
```
